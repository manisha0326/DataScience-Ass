```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Data
X = np.array([8, 15, 27, 34, 46, 52, 64, 71, 83, 95])
Y = np.array([10, 18, 30, 36, 48, 55, 67, 74, 85, 98])

# Mean, Median
print("Mean of X:", np.mean(X))
print("Median of X:", np.median(X))

# Mode using Pandas
X_series = pd.Series(X)
mode_X = X_series.mode()
if len(mode_X) == len(X):  # all values are unique
    print("Mode of X: No mode (all values are unique)")
else:
    print("Mode of X:", list(mode_X))

# Variance & Standard Deviation
print("Variance of X:", np.var(X))
print("Standard Deviation of X:", np.std(X))

# Covariance
print("Covariance X,Y:\n", np.cov(X, Y))

# Correlation
print("Correlation coefficient:", np.corrcoef(X, Y))
```

# Histogram of x

```python
plt.hist(X, bins=5, edgecolor='black')
plt.xlabel("X values")
plt.ylabel("Frequency")
plt.title("Histogram of X")
plt.show()
```

# Box plot of x

```python
plt.boxplot(X)
plt.title("Box Plot of X")
plt.ylabel("X values")
plt.show()
```

# Scatter plot of x and y

```python
plt.scatter(X, Y)
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Scatter Plot of X vs Y")
plt.show()
```

# Correlation matrix

```python
df = pd.DataFrame({'X': X, 'Y': Y})
print(df.corr())
```
